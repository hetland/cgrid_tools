# Simple Focused Channel

Generates a 20×80 rectangular channel with a focus transform applied along the xi
direction near the midpoint. Compares three solver modes (transfinite, Laplacian,
elliptic) and prints orthogonality statistics for each.

```python
import numpy as np

import cgrid_tools
from cgrid_tools.gridgen import BoundaryDefinition, Focus, GridGenerator, SolverMode

boundary = BoundaryDefinition(
    x=np.array([0.0, 20.0, 20.0, 0.0]),
    y=np.array([0.0, 0.0, 4.0, 4.0]),
    beta=np.array([1.0, 1.0, 1.0, 1.0]),
)
focus = Focus()
focus.add_focus(pos=0.5, axis="x", factor=3.0, extent=0.1)
shape = (21, 81)

print(f"Boundary: {len(boundary.x) - 1} vertices (closed)")
print(f"Shape: {shape}")
print()

grid = None
for mode in (SolverMode.TRANSFINITE, SolverMode.LAPLACIAN, SolverMode.ELLIPTIC):
    grid = GridGenerator(
        mode=mode,
        focus=focus,
        smoothing_iterations=200,
        elliptic_relaxation=1.3,
        elliptic_tolerance=1e-8,
    ).generate(boundary, shape=shape)
    o = cgrid_tools.orthogonality_summary(grid.x, grid.y)
    print(
        f"{mode.value:12s}: mean {np.degrees(o['mean_abs']):.3f}°"
        f"  max {np.degrees(o['max_abs']):.3f}°"
    )
```
