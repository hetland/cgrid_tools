```python
from matplotlib import pyplot as plt

from cgrid_tools.gridgen import BoundaryDefinition, CRDTOptions, Focus, GridGenerator, SolverMode
```

```python
x =    [2, 0, 0, 1, 1.75]
y =    [1, 1, 0, 0, 0.3]
beta = [1, 1, 1, 1, 0]

boundary = BoundaryDefinition(x=x, y=y, beta=beta)
focus = Focus()
focus.add_focus(0.5, axis="y", factor=3.0, extent=0.25)

result = GridGenerator(mode=SolverMode.ELLIPTIC, focus=focus).generate(
    boundary,
    shape=(21, 31),
)
```

```python
plt.plot(result.x, result.y, "k-")
plt.plot(result.x.T, result.y.T, "k-");
```

```python
x =    [3, 2,  2, 1, 1,  2, 2, 3]
y =    [4, 4,  3, 4, 3,  2, 1, 1]
beta = [1, 1, -1, 1, 1, -1, 1, 1]

boundary = BoundaryDefinition(x=x, y=y, beta=beta)
focus = Focus()
focus.add_focus(0.5, axis="y", factor=3.0, extent=0.25)
focus.add_focus(0.5, axis="x", factor=3.0, extent=0.25)

options = CRDTOptions(image_beta_blend=0.0)

grid = GridGenerator(
    mode=SolverMode.GRIDGEN_CRDT,
    crdt_options=options,
).generate(boundary, shape=(61, 41))

# result = GridGenerator(mode=SolverMode.GRIDGEN_CRDT).generate(
#     boundary,
#     shape=(100, 200),
# )
plt.plot(grid.x, grid.y, "k-")
plt.plot(grid.x.T, grid.y.T, "k-")
plt.plot(x, y, 'ro')
```
